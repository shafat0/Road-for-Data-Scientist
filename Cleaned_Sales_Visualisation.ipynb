```python
# Lets Import our data first:-
sales = pd.read_csv('/content/drive/MyDrive/sales_data_cleaned_drive.csv')
sales
```

```python
sales['total_sale'].plot(kind='line',
                         figsize=(20,10),
                          title = 'sales over time')
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()
plt.show()
```

* Clearly it is shown that the sales over time is maximum revenue between 900 and 1000 month.

```python
fig, ax = plt.subplots(figsize=(50,10))
sales.plot(kind='scatter',
           x = 'sale_date',
           y = 'sales_amount',
           c = 'total_sale',
           ax = ax)
```
