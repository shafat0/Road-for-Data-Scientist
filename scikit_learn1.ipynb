# Option 2:- Fill missing values with Scikit learn.

```python
car_sales_missing = pd.read_csv('/content/drive/MyDrive/car-sales-extended-missing-data.csv')
car_sales_missing
```

```python
# Let's check how many missing values do we have
car_sales_missing.isna().sum()
```

```python
# Now lets rid of the rows which don't have the 'Price' column value
car_sales_missing.dropna(subset = ['Price'], inplace = True)
```

```python
car_sales_missing.isna().sum()
```

```python
# Split into x and y
x = car_sales_missing.drop('Price', axis = 1)
y = car_sales_missing['Price']
```

```python
## Fill the missing values with the scikit learn
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
#1> Fill categorical values with "missing" and numerical values with mean
cat_imputer = SimpleImputer(strategy = 'constant', fill_value = 'missing')
door_imputer = SimpleImputer(strategy = 'constant', fill_value = 4)
num_imputer = SimpleImputer(strategy = 'mean')
#2> Define column
cat_feature = ['Make', 'Colour']
door_feature = ['Doors']
num_feature = ['Odometer (KM)']
#3> Create an imputer (Something that fills missing values)
imputer = ColumnTransformer([
    ('cat_imputer', cat_imputer,cat_feature),
    ('door_imputer', door_imputer,door_feature),
    ('num_imputer', num_imputer,num_feature)
])
#4> Transform the data
filled_x = imputer.fit_transform(x)
filled_x
```

```python
## Let's check now:-
car_sales_filled = pd.DataFrame(filled_x, columns = ['Make', 'Colour', 'Doors', 'Odometer (KM)'])
car_sales_filled
```

```python
car_sales_filled.isna().sum()
```

```python
## Let's convert the non-numerical data into numerical:-
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
categorical_features = ["Make", "Colour", "Doors"]
one_hot = OneHotEncoder()
transformer = ColumnTransformer([("one_hot",
                                  one_hot,
                                  categorical_features)],
                                remainder = 'passthrough')
transformed_x = transformer.fit_transform(car_sales_filled)
transformed_x
```

```python
pd.DataFrame(transformed_x)
```

```python
## Now we have got our data as numbers and filled the missing values
## Now let's fit the model
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(transformed_x,y,test_size = (0.2))
model = RandomForestRegressor()
model.fit(x_train,y_train)
model.score(x_test,y_test)
```

```python
### 2.1:- Picking a machine learning model for a regression problem:-
## Let's use the california housing dataset.
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
housing
```

```python
housing_df = pd.DataFrame(housing['data'],columns = housing["feature_names"])
housing_df
```

```python
housing_df["MedHouseVal"] = housing["target"]
housing_df
```

```python
housing_df = housing_df.drop("MedHouseVal",axis = 1)
```

```python
housing_df
```

```python
housing_df['target'] = housing['target']
```

```python
housing_df
```

```python
## import algorithm/estimator
from sklearn.linear_model import Ridge


## setup random seed
np.random.seed(42)
# create the data
x=housing_df.drop('target',axis=1)
y=housing_df['target'] ## Median house price in $100,000s

## Split into train and test sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)

# Instantiate and fit the model ( on the training set)
model = Ridge()
model.fit(x_train,y_train)

# Check the score of the model ( on the test set)
model.score(x_test,y_test)
```

```python
from sklearn import svm

## setup random seed
np.random.seed(42)
# create the data
x=housing_df.drop('target',axis=1)
y=housing_df['target'] ## Median house price in $100,000s

## Split into train and test sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)

# Instantiate and fit the model ( on the training set)
model = svm.SVR()
model.fit(x_train,y_train)

# Check the score of the model ( on the test set)
model.score(x_test,y_test)
```

```python
from sklearn import linear_model
## setup random seed
np.random.seed(42)
# create the data
x=housing_df.drop('target',axis=1)
y=housing_df['target'] ## Median house price in $100,000s

## Split into train and test sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)

# Instantiate and fit the model ( on the training set)
model = linear_model.LinearRegression()
model.fit(x_train,y_train)

# Check the score of the model ( on the test set)
model.score(x_test,y_test)
```

sklearn ensemble model can be found here:-https://scikit-learn.org/1.3/modules/ensemble.html
