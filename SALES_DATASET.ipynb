# SETP 1:-  Let's load the data using "pd.read_csv()" function.

```python
# Let's load the data using "pd.read_csv()" function.
sales = pd.read_csv('/content/drive/MyDrive/SALES_DATA.csv')
sales
```

```python
print("Original Data:")
print(sales.head(10))
print("\nInfo:")
print(sales.info())
```

# SETP 2:- Detect missing values using "df.isull().sum()" function.

```python
sales.isnull()
```

```python
sales.isnull().sum()
```

# STEP 3:- Now we will Replace all missing values in numeric columns with the mean of that column.

```python
# Handling numeric missing values
sales.fillna(sales.mean(numeric_only=True), inplace=True)
```

```python
# Handling categorical values
sales.fillna("unknown",inplace = True)
```

```python
sales
```

# Remove Duplicates:-

```python
sales.duplicated()
```

```python
sales.duplicated().sum()
```

```python
sales.drop_duplicates(inplace=True)
```

```python
sales
```

# SETP 5:-Rename columns to lowercase and replace spaces with underscores.

```python
sales.columns = [col.lower().replace(' ','_') for col in sales.columns]
```

```python
print('\n Columns after renaming:-')
print(sales.columns)
```

# STEP 6:- Summarize the cleaned data:-

```python
print('\n Summary statistics:-')
print(sales.describe())
```

```python
sales
```

```python
print('\n Total quantity  and average per region:-')
region_summary = sales.groupby('region').agg({'quantity_sold':'sum','unit_price':'mean'})
print(region_summary)
```

```python
print('\n Top 3 products by quantity sold:')
product_summary = sales.groupby('product_id')['quantity_sold'].sum().sort_values(ascending=False).head(3)
print(product_summary)
```

```python
# BONUS = Create total_sale column
sales['total_sale']=sales['quantity_sold']*sales['unit_price']
sales
```

```python
# Here we will check the data with highest total sales:
print('\n Date with highest total sales:')
date_sales = sales.groupby('sale_date')['total_sale'].sum()
print(date_sales.sort_values(ascending=False).head(1))
```
